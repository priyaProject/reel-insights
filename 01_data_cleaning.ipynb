```python
import numpy as np

movies_clean["budget_clean"] = movies_clean["budget"].replace(0, np.nan)

movies_clean["revenue_clean"] = movies_clean["revenue"].replace(0, np.nan)

print("Missing budget after cleaning:", movies_clean["budget_clean"].isna().sum())

print("Missing revenue after cleaning:", movies_clean["revenue_clean"].isna().sum())
```

```python
movies_clean["release_date"] = pd.to_datetime(
    movies_clean["release_date"],
    format="%Y-%m-%d",
    errors="coerce"
)

print("Release date datatype:", movies_clean["release_date"].dtype)

print("Invalid release dates:", movies_clean["release_date"].isna().sum())
```

```python
movies_clean["release_year"] = movies_clean["release_date"].dt.year

movies_clean["release_month"] = movies_clean["release_date"].dt.month

movies_clean["profit"] = (
    movies_clean["revenue_clean"] - movies_clean["budget_clean"]
)

movies_clean["roi"] = (
    (movies_clean["revenue_clean"] - movies_clean["budget_clean"])
    / movies_clean["budget_clean"]
) * 100

display(
    movies_clean[
        ["title", "release_year", "release_month",
         "budget_clean", "revenue_clean", "profit", "roi"]
    ].head(10)
)
```

```python
import sqlite3
from pathlib import Path

# Create database in the project root folder
db_path = Path("reel_insights.db")

# Avoid accidentally overwriting an existing database
if db_path.exists():
    raise FileExistsError(
        "reel_insights.db already exists. Do not overwrite it."
    )

conn = sqlite3.connect(db_path)

# Enable foreign key validation
conn.execute("PRAGMA foreign_keys = ON")

# Create Movies table
conn.execute("""
CREATE TABLE movies (
    movie_id INTEGER PRIMARY KEY,
    title TEXT,
    original_language TEXT,
    release_date TEXT,
    runtime INTEGER,
    budget REAL,
    revenue REAL,
    popularity REAL,
    vote_average REAL,
    vote_count INTEGER,
    status TEXT,
    overview TEXT,
    budget_clean REAL,
    revenue_clean REAL,
    release_year INTEGER,
    release_month INTEGER,
    profit REAL,
    roi REAL
)
""")

# Create Genres table
conn.execute("""
CREATE TABLE genres (
    genre_id INTEGER PRIMARY KEY,
    genre_name TEXT
)
""")

# Create Movie Genres table
conn.execute("""
CREATE TABLE movie_genres (
    movie_id INTEGER,
    genre_id INTEGER,
    PRIMARY KEY (movie_id, genre_id),
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id),
    FOREIGN KEY (genre_id) REFERENCES genres(genre_id)
)
""")

# Create Cast table
conn.execute("""
CREATE TABLE "cast" (
    cast_row_id INTEGER PRIMARY KEY,
    movie_id INTEGER,
    person_id INTEGER,
    actor_name TEXT,
    character_name TEXT,
    cast_order INTEGER,
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id)
)
""")

# Create Crew table
conn.execute("""
CREATE TABLE crew (
    crew_row_id INTEGER PRIMARY KEY,
    movie_id INTEGER,
    person_id INTEGER,
    person_name TEXT,
    job TEXT,
    department TEXT,
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id)
)
""")

# Create Movie Keywords table
conn.execute("""
CREATE TABLE movie_keywords (
    movie_id INTEGER,
    keyword_id INTEGER,
    keyword_name TEXT,
    PRIMARY KEY (movie_id, keyword_id),
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id)
)
""")

conn.commit()

print("All six SQLite tables created successfully!")
```

```python
import sqlite3
from pathlib import Path

db_path = Path("reel_insights.db")

conn = sqlite3.connect(db_path)

conn.execute("PRAGMA foreign_keys = ON")

print("Database connected successfully!")
print("Database location:", db_path.resolve())
```

```python
print(conn.execute("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
""").fetchall())
```

```python
movies_clean.to_sql("movies", conn, if_exists="append", index=False)

genres_clean.to_sql("genres", conn, if_exists="append", index=False)

movie_genres_clean.to_sql("movie_genres", conn, if_exists="append", index=False)

cast_clean.to_sql("cast", conn, if_exists="append", index=False)

crew_clean.to_sql("crew", conn, if_exists="append", index=False)

movie_keywords_clean.to_sql("movie_keywords", conn, if_exists="append", index=False)

conn.commit()

print("All six datasets inserted successfully!")
```

```python
tables = [
    "movies",
    "genres",
    "movie_genres",
    "cast",
    "crew",
    "movie_keywords"
]

for table in tables:
    result = conn.execute(
        f'SELECT COUNT(*) FROM "{table}"'
    ).fetchone()

    print(f"{table}: {result[0]} records")
```
