```python
import sqlite3
import pandas as pd
import sys

print("Python Path:", sys.executable)
print("Pandas Version:", pd.__version__)

conn = sqlite3.connect("reel_insights.db")

print("Connected to Reel Insights database!")
```

```python
query = """SELECT COUNT(*) AS total_movies, ROUND(AVG(vote_average), 2) AS average_rating
FROM movies;"""
result = pd.read_sql_query(query, conn)

display(result)
```

```python
query = """SELECT DISTINCT m.movie_id, m.title, m.release_year, g.genre_name FROM movies m JOIN movie_genres mg ON m.movie_id = mg.movie_id JOIN genres g ON mg.genre_id = g.genre_id WHERE g.genre_name = 'Action' AND m.release_year > 2015
ORDER BY m.release_year DESC;"""
action_movies = pd.read_sql_query(query, conn)

display(action_movies)
```

```python
query = """SELECT DISTINCT m.movie_id, m.title, m.release_year, g.genre_name FROM movies m
JOIN movie_genres mg ON m.movie_id = mg.movie_id JOIN genres g ON mg.genre_id = g.genre_id
WHERE g.genre_name = 'Action' AND m.release_year > 2015 ORDER BY m.release_year DESC;"""
action_movies = pd.read_sql_query(query, conn)
display(action_movies)

print("Total Action movies released after 2015:", len(action_movies))
```

```python
query = """SELECT m.movie_id, m.title, GROUP_CONCAT(DISTINCT g.genre_name) AS genres, m.revenue
FROM movies m JOIN movie_genres mg ON m.movie_id = mg.movie_id JOIN genres g ON mg.genre_id = g.genre_id
WHERE m.revenue_clean IS NOT NULL GROUP BY m.movie_id, m.title, m.revenue
ORDER BY m.revenue DESC LIMIT 10;"""
top_grossing_movies = pd.read_sql_query(query, conn)

display(top_grossing_movies)
```

```python
query = """SELECT g.genre_name, COUNT(DISTINCT m.movie_id) AS total_movies, ROUND(AVG(m.budget_clean), 2) AS average_budget,
  ROUND(AVG(m.revenue_clean), 2) AS average_revenue FROM movies m
JOIN movie_genres mg ON m.movie_id = mg.movie_id JOIN genres g ON mg.genre_id = g.genre_id
GROUP BY g.genre_name ORDER BY average_revenue DESC;"""

genre_financial_analysis = pd.read_sql_query(query, conn)

display(genre_financial_analysis)
```

```python
query = """SELECT person_id, actor_name, COUNT(DISTINCT movie_id) AS total_movies FROM "cast"
GROUP BY person_id, actor_name ORDER BY total_movies DESC LIMIT 10;"""
top_actors = pd.read_sql_query(query, conn)

display(top_actors)
```

```python
query = """SELECT person_id, person_name AS director_name,  COUNT(DISTINCT movie_id) AS total_movies
FROM crew WHERE job = 'Director' GROUP BY person_id, person_name HAVING COUNT(DISTINCT movie_id) > 3
ORDER BY total_movies DESC, director_name;"""

directors = pd.read_sql_query(query, conn)

display(directors)

print("Total directors with more than 3 movies:", len(directors))
```

```python
query = """SELECT keyword_id, keyword_name, COUNT(DISTINCT movie_id) AS movie_count
FROM movie_keywords GROUP BY keyword_id, keyword_name ORDER BY movie_count DESC, keyword_name
LIMIT 10;"""

top_keywords = pd.read_sql_query(query, conn)

display(top_keywords)
```

```python
query = """SELECT movie_id, title, budget_clean, revenue_clean FROM movies
WHERE budget_clean > ( SELECT AVG(budget_clean) FROM movies
) AND revenue_clean < ( SELECT AVG(revenue_clean) FROM movies)
ORDER BY budget_clean DESC;"""
low_performing_movies = pd.read_sql_query(query, conn)

display(low_performing_movies)

print("Total movies matching this condition:", len(low_performing_movies))
```

```python
query = """SELECT DISTINCT c.person_id, c.actor_name
FROM "cast" c JOIN crew cr ON c.movie_id = cr.movie_id
WHERE cr.job = 'Director' AND LOWER(TRIM(cr.person_name)) = 'christopher nolan'
ORDER BY c.actor_name;"""
nolan_actors = pd.read_sql_query(query, conn)

display(nolan_actors)

print("Total actors:", len(nolan_actors))
```

```python
query = """SELECT  g.genre_name, COUNT(DISTINCT m.movie_id) AS total_movies, ROUND(AVG(m.vote_average), 2) AS average_rating
FROM movies m JOIN movie_genres mg ON m.movie_id = mg.movie_id
JOIN genres g ON mg.genre_id = g.genre_id
GROUP BY g.genre_id, g.genre_name HAVING COUNT(DISTINCT m.movie_id) >= 20
ORDER BY average_rating DESC LIMIT 1;"""

highest_rated_genre = pd.read_sql_query(query, conn)

display(highest_rated_genre)
```

```python
query = """SELECT m.movie_id, m.title, COUNT(DISTINCT mk.keyword_id) AS keyword_count
FROM movies m JOIN movie_keywords mk ON m.movie_id = mk.movie_id GROUP BY m.movie_id, m.title
ORDER BY keyword_count DESC, m.title LIMIT 10;"""

top_keyword_movies = pd.read_sql_query(query, conn)

display(top_keyword_movies)
```
